```python
df["diagnosis_icd10_list"] = (
    df["diagnosis_icd10_list"]
    .astype(str)
    .str.replace(r'^[^A-Z0-9]+', '', regex=True)  # ลบหน้าค่า
    .str.replace(r'[^A-Z0-9,]', '', regex=True)   # ลบตัวแปลกอื่นๆ
)
```

```python
# ✅ เงื่อนไขใหม่:
# - ถ้าเป็นกลุ่ม J44*, J45*, H10*, L50*, I21*, I22*, I24* → ให้ match จาก “3 ตัวหน้า”
#   (เช่น J450/J451 → อยู่ในกลุ่ม J45)
# - ยกเว้น L309 → ต้อง match “L309 เท่านั้น” (ไม่ใช่ L30*)

prefixes = ["J44", "J45", "H10", "L50", "I21", "I22", "I24"]
exact = {"L309"}

def clean_token(tok: str) -> str:
    tok = str(tok).strip().upper()
    tok = re.sub(r"^[^A-Z0-9]+", "", tok)  # ตัดอักขระแปลกๆ หน้ารหัส เช่น "๋J449"
    tok = re.sub(r"[^A-Z0-9]+$", "", tok)  # ตัดท้ายที่ไม่ใช่ตัวอักษร/ตัวเลข
    return tok

def extract_targets(icd):
    if pd.isna(icd):
        return None
    out = []
    for raw in str(icd).split(","):
        tok = clean_token(raw)
        if not tok:
            continue

        # L309 ต้องตรงตัวเท่านั้น
        if tok in exact:
            if tok not in out:
                out.append(tok)
            continue

        # กลุ่มอื่น match จาก 3 ตัวหน้า
        if tok[:3] in prefixes:
            if tok not in out:
                out.append(tok)  # เก็บ “รหัสเต็ม” ที่พบ เช่น J450, H103
    return ",".join(out) if out else None

# เพิ่มคอลัมน์ใหม่
df["icd_target"] = df["diagnosis_icd10_list"].apply(extract_targets)

```

```python
mapping = {
    "J44": "Chronic obstructive pulmonary disease",
    "J45": "Acute Asthma",
    "H10": "กลุ่มโรคตาอักเสบ",
    "L50": "กลุ่มโรคผิวหนังอักเสบ",
    "L309": "กลุ่มโรคผิวหนังอักเสบ",
    "I21": "Acute Ischemic heart disease",
    "I22": "Acute Ischemic heart disease",
    "I24": "Acute Ischemic heart disease",
}

def map_disease(icd):
    for k, v in mapping.items():
        if k in icd:
            return v
    return None

df2["Disease"] = df2["icd_target"].apply(map_disease)
```

```python
# สร้าง dictionary สำหรับจัดกลุ่มโรค
disease_group_map = {
    "Chronic obstructive pulmonary disease": "กลุ่มโรคทางเดินหายใจ",
    "Acute Asthma": "กลุ่มโรคทางเดินหายใจ",
    "Acute Ischemic heart disease": "กลุ่มโรคหัวใจและหลอดเลือด",
    "กลุ่มโรคตาอักเสบ": "กลุ่มโรคตาอักเสบ",
    "กลุ่มโรคผิวหนังอักเสบ": "กลุ่มโรคผิวหนังอักเสบ"
}

# เพิ่มคอลัมน์ใหม่
df_patient["disease group"] = df_patient["Disease"].map(disease_group_map)

# ถ้ามีโรคอื่นที่ไม่ได้อยู่ใน mapping ให้กำหนดเป็น "อื่น ๆ"
df_patient["disease group"] = df_patient["disease group"].fillna("อื่น ๆ")

df_patient.head()
```

```python
# 1️⃣ กรองข้อมูลเฉพาะปี 2021–2025
df_base = df_patient[(df_patient["year"] >= 2021) & (df_patient["year"] <= 2025)].copy()

# 2️⃣ คำนวณ median และ IQR แยกตาม Disease + province
stats = (
    df_base
    .groupby(["Disease", "province"])["patient_count"]
    .agg(
        median="median",
        Q1=lambda x: x.quantile(0.25),
        Q3=lambda x: x.quantile(0.75)
    )
    .reset_index()
)

# 3️⃣ คำนวณ IQR และ median + 1.5*IQR
stats["IQR"] = stats["Q3"] - stats["Q1"]
stats["median+1.5IQR"] = stats["median"] + 1.5 * stats["IQR"]

# 4️⃣ เลือกเฉพาะคอลัมน์ที่ต้องใช้
stats = stats[["Disease", "province", "median+1.5IQR"]]

# 5️⃣ merge กลับเข้า df หลัก
df_patient = df_patient.merge(stats, on=["Disease", "province"], how="left")

df_patient.head()
```

```python
# ฟังก์ชันคำนวณ median + 1.5*IQR
def median_plus_1_5iqr(x):
    q1 = x.quantile(0.25)
    q3 = x.quantile(0.75)
    iqr = q3 - q1
    median = x.median()
    return median + (1.5 * iqr)

# คำนวณแบบ groupby แล้วใส่กลับเข้า df
df_patient["median+1.5IQR"] = (
    df_patient.groupby(["Disease", "province"])["patient_count"]
      .transform(median_plus_1_5iqr)
)

df_patient.head()
```

```python
# 1) แปลง date ให้เป็น datetime (รูปแบบ dd/mm/YYYY ตามภาพ)
pm["date"] = pd.to_datetime(pm["date"], format="%d/%m/%Y", errors="coerce")

# 2) สร้าง ISO year/week และวันที่วันจันทร์ของสัปดาห์นั้น
iso = pm["date"].dt.isocalendar()
pm["year"] = iso.year
pm["week"] = iso.week
pm["wk_year"] = pd.to_datetime(pm["year"].astype(str) + pm["week"].astype(str) + "1", format="%G%V%u")

# 3) คำนวณค่าเฉลี่ยรายจังหวัด-รายสัปดาห์
pm_weekly = (
    pm.groupby(["province", "year", "week", "wk_year"], as_index=False)
      .agg(avg_pm25=("pm25", "mean"))
      .sort_values(["province", "year", "week"])
)

# ทางเลือก: ปัดทศนิยมให้อ่านง่าย
pm_weekly["avg_pm25"] = pm_weekly["avg_pm25"].round(2)
```
